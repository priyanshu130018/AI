# Embeddings / Vector

```python
sentences = [
    "Python is a programming language.",
    "Python is used to write computer programs.",
    "RAG retrieves information before generating an answer.",
    "Dogs are animals.",
    "The weather is sunny today."
]
```

```python
import numpy as np
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

embeddings = model.encode(sentences)

print("Type:", type(embeddings))
print("Shape:", embeddings.shape)
```

```python
for sentence, embedding in zip(sentences, embeddings):
    print(f"Sentence : {sentence}")
    print(f"Embedding : [{embedding[0]:.4f}, {embedding[1]:.4f}, {embedding[2]:.4f}, ...]")
    print(f"Embedding Length : {len(embedding)}")
    print("-" * 50)
```

# Similarity

## Cosine Similarity

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity_matrix = cosine_similarity(embeddings)

print("\nSimilarity Matrix")
print(similarity_matrix)
```

```python
for i in range(len(sentences)):
    for j in range(i + 1, len(sentences)):

        similarity = cosine_similarity(
            [embeddings[i]],
            [embeddings[j]]
        )[0][0]

        print(f"{sentences[i]} ↔ {sentences[j]}")
        print(f"Similarity : {similarity:.4f}")
        print("-" * 50)
```

```python
from sentence_transformers import util

for i in range(len(sentences)):
    for j in range(i + 1, len(sentences)):

        similarities = util.cos_sim(
            embeddings[i],
            embeddings[j]
        ).item()

        print(f"{sentences[i]} ↔ {sentences[j]}")
        print(f"Similarity : {similarities:.4f}")
        print("-" * 50)
```

## Manual Cosine Similarity

```python
def cosine_similarity_manual(a, b):

    dot = np.dot(a, b)

    magnitude_a = np.linalg.norm(a)
    magnitude_b = np.linalg.norm(b)

    return dot / (magnitude_a * magnitude_b)
```

```python
for i in range(len(sentences)):
    for j in range(i + 1, len(sentences)):

        score = cosine_similarity_manual(
            embeddings[i],
            embeddings[j]
        )

        print(f"{sentences[i]} ↔ {sentences[j]}")
        print(f"Similarity : {score:.4f}")
        print("-" * 50)
```
