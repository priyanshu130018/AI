# 27 - MCP (Model Context Protocol)

MCP is a standard way for AI applications to connect to external capabilities.

Current MCP Python SDK v2 documentation describes:
- MCP servers
- tools
- resources
- prompts
- clients
- standard transports such as stdio, Streamable HTTP, and SSE. citeturn983125view0

This makes MCP an integration protocol, not an agent framework by itself.

## 1. MCP mental model

```text
AI application / host
        ↓
      MCP client
        ↓
    MCP server
    ├─ tools
    ├─ resources
    └─ prompts
```

Tools let the model request actions. Resources expose data. Prompts are reusable templates surfaced to users/clients. citeturn520106search4turn520106search0

## 2. Current Python SDK v2 server

The current official Python SDK v2 docs show an `MCPServer` with decorators for tools and resources.

````python
# Save this example as server.py to run it outside the notebook.
server_code = '''from mcp.server import MCPServer

mcp = MCPServer("Demo")

@mcp.tool()
def add(a: int, b: int) -> int:
    "Add two numbers."
    return a + b

@mcp.resource("greeting://{name}")
def greeting(name: str) -> str:
    "Greet someone by name."
    return f"Hello, {name}!"
'''

print(server_code)
````

## 5. MCP vs function calling

Function calling:
```text
model ↔ application tool schema
```

MCP:
```text
application ↔ standardized external capability server
```

MCP can standardize how capabilities are exposed across hosts and clients.

## 6. Security

MCP is not a magic trust boundary.

Apply:
- authentication
- authorization
- least privilege
- tool allowlists
- input validation
- output validation
- network controls
- audit logs
