# 20 - Planning

Planning adds an explicit plan before or during action.

Example:

```text
Goal
 ↓
Plan
 ├─ research
 ├─ compare
 ├─ calculate
 └─ verify
       ↓
   execute
```

Planning is useful for tasks where the agent must manage dependencies, subgoals, or long sequences.

## 1. Plan vs direct execution

Direct:
```text
goal → action → answer
```

Plan:
```text
goal → subgoals → actions → verification → answer
```

Planning can reduce aimless tool usage, but planning itself costs time/tokens and can become wrong. Evaluate whether it actually improves task success.

## 2. Structured plan

Use a typed plan when downstream code must consume it.

````python
from pydantic import BaseModel

class Plan(BaseModel):
    goal: str
    steps: list[str]

plan = Plan(
    goal='Compare two RAG systems',
    steps=[
        'Define evaluation questions',
        'Run baseline',
        'Run advanced RAG',
        'Compare metrics',
        'Write recommendation',
    ],
)

print(plan.model_dump())
````

## 3. Plan execution

Each plan step can become a graph node, tool action, or subtask.

````python
def execute_plan(plan):
    results = []
    for step in plan.steps:
        results.append(f'DONE: {step}')
    return results

print(execute_plan(plan))
````

## 4. Replanning

A strong planner can update its plan when observations invalidate assumptions.

```text
plan
 ↓
execute
 ↓
unexpected result
 ↓
replan
 ↓
continue
```

````python
state = {
    'plan': ['search docs', 'summarize'],
    'observation': 'docs missing',
}

if 'missing' in state['observation']:
    state['plan'] = ['find another source', 'search docs again', 'summarize']

print(state)
````

## 5. Plan-and-execute vs reactive

Reactive agent:
- decides one action at a time.

Plan-and-execute:
- creates a multi-step plan first.

Choose based on task characteristics. Simple tasks often do not need heavy planning.

## 6. Evaluation and practice

Evaluate:
- task completion
- plan accuracy
- unnecessary steps
- replan success
- cost
- latency

Practice: make a research plan, deliberately fail one step, and replan.

## Evaluation

Planning should be judged by outcomes, not by how impressive the plan text looks.

Improve by:
- limiting plan length
- using typed steps
- verifying assumptions
- allowing replan
- stopping early when the goal is reached.
