# CHUNKING TECHNIQUES

```python
text = """
Artificial Intelligence is transforming industries.
Machine Learning is a subset of AI.
Deep Learning uses neural networks.
Random Forest combines many decision trees.
Decision Trees are supervised learning algorithms.
Pizza is an Italian dish.
Burger is popular fast food.
"""
```

## LangChain Recursive Chunking

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=50,
    chunk_overlap=10,
    separators=[
        "\n\n",
        "\n",
        ". ",
        " ",
        ""
    ]
)

chunks = splitter.split_text(text)

print(f"Total Chunks : {len(chunks)}\n")

for i, chunk in enumerate(chunks, 1):
    print(f"Chunk {i}")
    print(chunk)
    print(f"Length : {len(chunk)}")
    print("-" * 50)
```

## Fixed Chunking

```python
def fixed_chunk(text, chunk_size=50):
    chunks = []
    for i in range(0, len(text), chunk_size):
        chunks.append(text[i:i + chunk_size])
    return chunks
```

```python
for i, chunk in enumerate(fixed_chunk(text), 1):
    print(f"Chunk {i}")
    print(chunk)
    print("-" * 50)
```

## Overlap Chunking

```python
def overlap_chunk(text, chunk_size=50, overlap=10):

    chunks = []
    start = 0

    while start < len(text):
        end = start + chunk_size
        chunks.append(text[start:end])
        start += chunk_size - overlap

    return chunks
```

```python
for i, chunk in enumerate(overlap_chunk(text), 1):
    print(f"Chunk {i}")
    print(chunk)
    print("-" * 50)
```

## Sentence Chunking

```python
def sentence_chunk(text, sentences_per_chunk=2):

    sentences = text.strip().split(". ")

    chunks = []

    for i in range(0, len(sentences), sentences_per_chunk):

        chunk = ". ".join(sentences[i:i + sentences_per_chunk])

        if not chunk.endswith("."):
            chunk += "."

        chunks.append(chunk)

    return chunks
```

```python
for i, chunk in enumerate(sentence_chunk(text), 1):
    print(f"Chunk {i}")
    print(chunk)
    print("-" * 50)
```

## Recursive Chunking

```python
def recursive_chunk(text, chunk_size, separators):

    if len(text) <= chunk_size:
        return [text]

    if not separators:
        return [text[i:i + chunk_size] for i in range(0, len(text), chunk_size)]

    separator = separators[0]

    if separator == "":
        return [text[i:i + chunk_size] for i in range(0, len(text), chunk_size)]

    parts = text.split(separator)

    chunks = []
    current = ""

    for part in parts:

        piece = part + separator

        if len(current) + len(piece) <= chunk_size:
            current += piece
        else:
            if current:
                chunks.append(current.strip())
            current = piece

    if current:
        chunks.append(current.strip())

    final_chunks = []

    for chunk in chunks:
        if len(chunk) > chunk_size:
            final_chunks.extend(recursive_chunk(chunk, chunk_size, separators[1:]))
        else:
            final_chunks.append(chunk)

    return final_chunks
```

```python
separators = [
    "\n\n",
    "\n",
    ". ",
    " ",
    ""
]

for i, chunk in enumerate(recursive_chunk(text, 60, separators), 1):
    print(f"Chunk {i}")
    print(chunk)
    print("-" * 50)
```

## Token Chunking

```python
def token_chunk(text, chunk_size=8):

    words = text.split()

    chunks = []

    for i in range(0, len(words), chunk_size):
        chunks.append(" ".join(words[i:i + chunk_size]))

    return chunks
```

```python
for i, chunk in enumerate(token_chunk(text), 1):
    print(f"Chunk {i}")
    print(chunk)
    print("-" * 50)
```

## Semantic Chunking

```python
def semantic_chunk():

    documents = [
        "Artificial Intelligence is transforming industries.",
        "Machine Learning is a subset of AI.",
        "Deep Learning uses neural networks.",
        "Random Forest is an ensemble algorithm.",
        "Decision Trees are supervised learning algorithms.",
        "Pizza is an Italian dish.",
        "Burger is popular fast food."
    ]

    chunks = [[], [], []]

    for sentence in documents:

        if any(word in sentence.lower() for word in ["artificial", "machine", "deep"]):
            chunks[0].append(sentence)

        elif any(word in sentence.lower() for word in ["forest", "tree"]):
            chunks[1].append(sentence)

        else:
            chunks[2].append(sentence)

    return chunks
```

```python
for i, chunk in enumerate(semantic_chunk(), 1):
    print(f"Semantic Chunk {i}")
    print("\n".join(chunk))
    print("-" * 50)
```
