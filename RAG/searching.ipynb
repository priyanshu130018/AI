# Searching

```python
documents = [
    "Artificial Intelligence",
    "Machine Learning",
    "Random Forest",
    "Deep Learning",
    "Pizza Recipe"
]
```

## FAISS Vector Search

```python
import faiss
import numpy as np
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
```

Create Embedding

```python
doc_embeddings = model.encode(documents)

query = "Explain Random Forest"
query_embedding = model.encode([query])
```

Create FAISS Index

```python
dimension = doc_embeddings.shape[1]
index = faiss.IndexFlatL2(dimension)
```

Add Vectors

```python
index.add(doc_embeddings.astype("float32"))
print("Total Vectors :", index.ntotal)
```

Vector Search

```python
top = 3
distances, indices = index.search(
    query_embedding.astype("float32"),
    top
)

print("\nTop Results\n")
for rank, idx in enumerate(indices[0], start=1):

    print(f"Rank {rank}")
    print(f"Document : {documents[idx]}")
    print(f"Distance : {distances[0][rank-1]:.4f}")
    print("-" * 50)
```

## Cosine Similarity Search

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

Create Embedding

```python
doc_embeddings = model.encode(documents)

query = "Explain Random Forest"
query_embedding = model.encode([query])
```

Cosine Similarity

```python
scores = cosine_similarity(query_embedding, doc_embeddings)

for doc, score in zip(documents, scores[0]):
    print(f"{doc:<30} : {score:.4f}")

best_index = np.argmax(scores)
```

```python
print("\nBest Match")
print(f"Document : {documents[best_index]}")
print(f"Score    : {scores[0][best_index]:.4f}")
```

## Euclidean Distance Search

```python
from sentence_transformers import SentenceTransformer
import numpy as np
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

Create Embedding

```python
doc_embeddings = model.encode(documents)

query = "Explain Random Forest"
query_embedding = model.encode([query])
```

Compare Eculideand distance

```python
distances = []

for doc, embedding in zip(documents, doc_embeddings):

    distance = np.linalg.norm(query_embedding[0] - embedding)
    distances.append(distance)
    print(f"{doc:<30} : {distance:.4f}")
```

```python
closest_index = np.argmin(distances)
print("\nClosest Match")
print(f"Document : {documents[closest_index]}")
print(f"Distance : {distances[closest_index]:.4f}")
```
